```python
import numpy as np
from ldpc.bp_decoder import BpDecoder
```

```python
pcm = np.array([[1, 1, 0],
                [0, 1, 1]], dtype=np.uint8)

decoder = BpDecoder(pcm, error_channel=[0.1, 0.1, 0.1], max_iter=10)

llrs = np.array([5.0, -1.0, 0.5], dtype=np.float64)  # your AWGN-derived LLRs
decoded = decoder.decode(llrs)
print(decoded)
```
