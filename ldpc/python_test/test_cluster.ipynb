```python
import numpy as np
from ldpc.bp_decoder import BpDecoder
```

```python
pcm = np.array([[1, 1, 0],
                [0, 1, 1]], dtype=np.uint8)

decoder = BpDecoder(pcm, error_channel=[0.1, 0.1, 0.1], schedule="cluster")
cluster = np.array([0], dtype=np.int32)


updated_llrs = decoder.decode_cluster(np.array([0.5, -1.2, 0.3], dtype=np.float64), cluster)
print(updated_llrs)
```
